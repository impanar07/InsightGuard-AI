```python
from pathlib import Path

import pandas as pd

from IPython.display import display

pd.set_option("display.max_columns", None)


def find_project_root(
    start_path: Path
) -> Path:

    start_path = start_path.resolve()

    for path in [start_path, *start_path.parents]:

        if (
            (path / "data").exists()
            and (path / "src").exists()
        ):
            return path

    raise FileNotFoundError(
        "Project root could not be found."
    )


PROJECT_ROOT = find_project_root(
    Path.cwd()
)

REPORT_PATH = (
    PROJECT_ROOT
    / "reports"
    / "data_quality"
)

print(f"Project root: {PROJECT_ROOT}")
print(f"Report path: {REPORT_PATH}")
```

```python
overall_score = pd.read_csv(
    REPORT_PATH / "overall_quality_score.csv"
)

display(overall_score)
```

```python
quality_scorecard = pd.read_csv(
    REPORT_PATH
    / "data_quality_scorecard.csv"
)

display(
    quality_scorecard.sort_values(
        "data_quality_score"
    )
)
```

```python
weakest_datasets = (
    quality_scorecard
    .sort_values(
        "data_quality_score"
    )
    .head(5)
)

display(weakest_datasets)
```

```python
weakest_datasets = (
    quality_scorecard
    .sort_values(
        "data_quality_score"
    )
    .head(5)
)

display(weakest_datasets)
```

```python
issues = pd.read_csv(
    REPORT_PATH
    / "detected_quality_issues.csv"
)

print(
    f"Total detected issues: "
    f"{len(issues):,}"
)

display(issues)
```

```python
severity_summary = (
    issues
    .groupby("severity")
    .size()
    .reset_index(
        name="issue_count"
    )
)

severity_order = [
    "critical",
    "high",
    "medium",
    "low",
    "none",
]

severity_summary["severity"] = pd.Categorical(
    severity_summary["severity"],
    categories=severity_order,
    ordered=True
)

severity_summary = (
    severity_summary
    .sort_values("severity")
)

display(severity_summary)
```

```python
dataset_issue_summary = (
    issues
    .groupby("dataset")
    .size()
    .reset_index(
        name="issue_count"
    )
    .sort_values(
        "issue_count",
        ascending=False
    )
)

display(dataset_issue_summary)
```

```python
completeness = pd.read_csv(
    REPORT_PATH
    / "completeness_report.csv"
)

missing_columns = (
    completeness[
        completeness["missing_count"] > 0
    ]
    .sort_values(
        "missing_count",
        ascending=False
    )
)

display(missing_columns)
```

```python
overall_quality_score = (
    overall_score[
        "overall_data_quality_score"
    ].iloc[0]
)

total_issues = len(issues)

critical_issues = (
    issues["severity"]
    .eq("critical")
    .sum()
)

high_issues = (
    issues["severity"]
    .eq("high")
    .sum()
)

print("=" * 80)
print("INSIGHTGUARD-AI — DATATRUST AI SUMMARY")
print("=" * 80)

print(
    f"\nOverall Quality Score: "
    f"{overall_quality_score:.2f}/100"
)

print(
    f"Total Detected Issues: "
    f"{total_issues:,}"
)

print(
    f"Critical Issues: "
    f"{critical_issues:,}"
)

print(
    f"High Severity Issues: "
    f"{high_issues:,}"
)

print("\nLowest Quality Datasets:")

print(
    quality_scorecard[
        [
            "dataset",
            "data_quality_score",
        ]
    ]
    .sort_values(
        "data_quality_score"
    )
    .head(5)
    .to_string(index=False)
)

print("=" * 80)
```
