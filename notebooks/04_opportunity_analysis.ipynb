```python
from pathlib import Path

import pandas as pd

from IPython.display import display

pd.set_option(
    "display.max_columns",
    None
)


def find_project_root(
    start_path: Path
) -> Path:

    start_path = start_path.resolve()

    for path in [
        start_path,
        *start_path.parents
    ]:

        if (
            (path / "data").exists()
            and (path / "src").exists()
        ):
            return path

    raise FileNotFoundError(
        "Project root could not be found."
    )


PROJECT_ROOT = find_project_root(
    Path.cwd()
)

REPORT_PATH = (
    PROJECT_ROOT
    / "reports"
    / "opportunity"
)

print(PROJECT_ROOT)
```

```python
all_opportunities = pd.read_csv(
    REPORT_PATH
    / "all_opportunities_ranked.csv"
)

display(
    all_opportunities.head(20)
)
```
